```python
import pandas as pd
import numpy as np
import random
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.columns = [
    "Customer_ID",
    "Subscription_Length",
    "Customer_Satisfaction",
    "Daily_Watch_Time",
    "Engagement_Rate",
    "Device",
    "Genre",
    "Region",
    "Payment_History",
    "Subscription_Plan",
    "Churn",
    "Support_Queries",
    "Age",
    "Monthly_Income",
    "Promotional_Offers",
    "Profiles_Created"
]
```

```python
data.dtypes
```

```python
for col in data.select_dtypes(include="object").columns:
    print(f"\n{col}")
    print(data[col].unique())
```

```python
for col in data.select_dtypes(include="object").columns:
    data[col] = data[col].str.strip()
```

```python
days = []

for index, row in data.iterrows():

    if row["Churn"] == "Yes":

        if row["Daily_Watch_Time"] < 2:
            days.append(random.randint(45, 90))

        else:
            days.append(random.randint(30, 60))

    else:

        if row["Daily_Watch_Time"] > 4:
            days.append(random.randint(0, 10))

        else:
            days.append(random.randint(10, 30))

data["Days_Since_Last_Activity"] = days
```
